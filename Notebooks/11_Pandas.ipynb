# **11. Pandas**



Pandas es un módulo de python para análisis de datos. Pandas ofrece una variedad de métodos para manipulación de datos que permite ejecutar tareas complejas con comandos simples (de una línea).

Mientras **NumPy** es un módulo construido alrededor de la estructura de una __matriz__ (ndarray), **Pandas** es un módulo construido alrededor de la estructura de un __DataFrame__.

Un **DataFrame** es esencialmente una tabla de datos, similar a una hoja de cálculo en Excel. Está diseñado para almacenar datos en forma de filas y columnas. Los **DataFrames** proporcionan una forma estructurada de organizar y analizar datos tabulares

Un __DataFrame__ se asemeja a una matriz, con algunas diferencias:

* En un **DataFrame**, las columnas tienen nombres de cabecera
* Diferentes tipos de datos (int, str, bool, entre otros) se permiten dentro del mismo __DataFrame__. Cada columna tiene su propio tipo de dato (dtype o type).
* Cada <font color="red">__columna__</font> representa una variable o una característica
* Cada <font color="blue">__fila__</font> representa una observación o un dato
* Tanto las filas como columnas tienen una <font color="purple">__etiqueta__</font>

En pandas, la diferencia entre una Serie (Series) y un DataFrame es similar a la diferencia entre un vector unidimensional y una matriz bidimensional en NumPy.
- __Series__: Estructura de datos unidimensional (similar a un vector), que contiene valores y un índice.
- __DataFrame__: Estructura de datos bidimensional (similar a una matriz), organizada en filas y columnas. Cada columna es una Series.

**¿Qué son las *Series?***

Cuando se usa un solo par de corchetes, se extrae una columna individual como un objeto de una sola dimensión.

* Es similar a una lista de Python con esteroides (tiene un índice) (su dimensión es 1D).
* Ideal cuando para hacer operaciones matemáticas simples, como `df['Muertos'].mean()` o `df['Muertos'].sum()`.

**¿Cuándo es necesario el usar el doble corchete?**

Cuando usas doble corchete, le estás pasando a Pandas una lista de columnas (aunque solo sea una). Esto obliga a mantener el formato de tabla.

* Una tabla completa que solo tiene una columna (su dimensión es 2D).

* Es necesario cuando se van a aplicar métodos que solo existen para DataFrames, o seleccionar varias columnas, por ejemplo `df[['Muertos', 'Confirmed']]`.

## **11.2.2.  Slicing usando `.loc` y `.iloc`**

Para usar correctamente slicing, es necesario diferenciar entre la **Etiqueta** y la **Posición** de las filas y columnas

### **Etiqueta vs. Posición**

* Index/Row Label (**Etiqueta**):

Es el nombre que tú le asignas (como 'zero', 'one'). Es lo que identificamos con el atributo .index.

* Index/Row Position (**Posición**):

Es el orden físico en la memoria, empezando siempre desde 0 hasta n-1, sin importar el nombre que tenga la fila.

![](https://miro.medium.com/max/720/1*bEUFOBKEvgZnmBbEPyK3tg.png)






### **`.loc`** vs **`.iloc`**

Los métodos `.loc` y `.iloc` se utilizan para hacer **Slicing**, es decir, seleccionar los Index/Row de interés.


* **Label-based** (`.loc`):

Se usa para seleccionar por el nombre (o etiqueta).

En los rangos (slicing) **incluye el último elemento**. Si pides 'zero':'two', te traerá tres filas: zero, one y two.

* **Position based** (`.iloc`):

Se usa para seleccionar por el número de posición (enteros).

Sigue la lógica estándar de Python **el último elemento está excluido**. Si pides 0:2, solo te traerá las posiciones 0 y 1.

![](https://miro.medium.com/max/720/1*dYtynwab99wnMqfgyPUd3w.png)

# **11.5. Grafico con dos ejes verticales**

````python
# 1. Seleccionar datos
df20 = df[['Muertos','Activos']].sort_values(by=['Activos'],ascending=False)[:30]

# 2. Crear la figura y el primer eje (ax1)
fig, ax1 = plt.subplots(figsize=(16, 3))

# 3. Graficar en el primer eje
color_1 = 'tab:red'
ax1.plot(df20['Muertos'],'-or', alpha=0.6)
ax1.set_ylabel('Cantidad de muertos', color=color_1)
ax1.tick_params(axis='y', labelcolor=color_1)
ax1.tick_params(axis='x', rotation=60)

# 4. Crear el segundo eje (ax2) que comparte el mismo eje X
ax2 = ax1.twinx()

# 5. Graficar en el segundo eje
color_2 = 'tab:blue'
ax2.plot(df20['Activos']/1e6,':1b')
ax2.set_ylabel('Casos activos (millones)', color=color_2)
ax2.tick_params(axis='y', labelcolor=color_2)

# 6. Mostrar el gráfico
plt.title('Gráfico con doble eje vertical en Matplotlib')
plt.show()
````

* __NOTA:__ También se puede hacer este tipo de gráficos con arreglos de NumPy

````python
# 1. Crear datos de ejemplo
x = np.arange(0, 10, 1)
y1 = x ** 2          # Datos para el eje Y izquierdo
y2 = x * 1000        # Datos para el eje Y derecho (escala mucho mayor)

# 2. Crear la figura y el primer eje (ax1)
fig, ax1 = plt.subplots(figsize=(8, 5))

# 3. Graficar en el primer eje
color_1 = 'tab:red'
ax1.plot(x, y1, color=color_1, marker='o')
ax1.set_xlabel('Eje X (Tiempo)')
ax1.set_ylabel('Eje Y Izquierdo ($x^2$)', color=color_1)
ax1.tick_params(axis='y', labelcolor=color_1)

# 4. Crear el segundo eje (ax2) que comparte el mismo eje X
ax2 = ax1.twinx()

# 5. Graficar en el segundo eje
color_2 = 'tab:blue'
ax2.plot(x, y2, color=color_2, marker='s')
ax2.set_ylabel('Eje Y Derecho (x * 1000)', color=color_2)
ax2.tick_params(axis='y', labelcolor=color_2)

# 6. Mostrar el gráfico
plt.title('Gráfico con doble eje vertical en Matplotlib')
plt.show()
````
