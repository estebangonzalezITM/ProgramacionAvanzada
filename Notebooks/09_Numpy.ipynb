### **9.2.1. Arreglos (arrays)**

Un arreglo (array) es una estructura de datos diseñada para almacenar una colección de elementos (generalmente números) **del mismo tipo** en posiciones de memoria contiguas.

<font color="blue" size=4>**Características fundamentales**</font>

* **Homogeneidad**

A diferencia de una lista de Python (que puede mezclar peras con manzanas), un arreglo exige que todos sus datos sean iguales (por ejemplo, que todos sean float64). Esto permite que la computadora procese los datos mucho más rápido.

* **Eficiencia de memoria**

Al estar "pegados" unos con otros en la RAM, el procesador puede acceder a ellos de forma secuencial sin saltos innecesarios, lo que optimiza el uso de la caché de tu CPU.

* **Dimensiones**

**1D:** Un vector (una fila de datos).

**2D:** Una matriz (filas y columnas).

**3D o más:** Tensores (útiles para datos volumétricos o series temporales).

En el estado inicial `A` tiene dimensiones `(3,)`, por tanto, es una simple fila de datos sin concepto de "columnas".

Al ejecutar `np.expand_dims(A, axis=1)`, lo que se hace es insertar una nueva dimensión en la posición del índice 1 (las columnas). Ahora es su tamaño es `(3,1)`.

Técnicamente, se acaba de transformar un vector (1D) en una matriz columna (2D).

* **Sumar listas** vs **Sumar arreglos**

```python
L1 = [1,2,3,4]
L2 = [5,3,2,19]
print(type(L1), type(L2))
```

```python
#Sumar estas dos listas
R = L1 + L2
print("Resultado:",R)
```

```python
A1 = np.array(L1)
A2 = np.array(L2)
print(A1)
print(A2)
```

```python
#Sumar estos dos arreglos
Rn = A1 + A2
print("Resultado:",Rn)
```

Si queremos que las listas se comporten como los arreglos, podemos crear una función que sume las listas como su fueran arreglos:

```python
def SumaLista(L1,L2):
    R = []
    for i in range(len(L1)):
        R.append(L1[i]+L2[i])

    return R
```

```python
SumaLista(L1,L2)
```

```python
SumaLista([3,5,2,1,2,4,6,2],[1,2,4,56,8,9,1,2])
```

Sin embargo, la función `SumaLista` **es mucho menos eficiente que la suma de arreglos**. Para demostrarlo usaremos dos vectores muy grandes y `%timeit`, que se utiliza para medir el tiempo de ejecución de una línea de código de forma precisa y estadística.

```python
Lista1 = list(range(1000000))
Lista2 = list(range(1000000))
```

```python
%%timeit
RL = SumaLista(Lista1,Lista2)
```

```python
Arr1 = np.array(Lista1)
Arr2 = np.array(Lista2)
```

```python
%%timeit
RA = Arr1 + Arr2
```

```python
RA == RL   # Comparación elemento a elemento
```

```python
np.all(RA == RL)    # Comparación entre los dos arreglos
```

### **Ejemplo**

La definición formal de un número aleatorio en un rango es: $(b-a)*rand + a$. Utilicemos esta definición para crear una matriz aleatoria cuyos elementos tengan valores entre 5 y 15.

```python
import random
M = np.zeros([3,5])

a = 5
b = 15

for i in range(M.shape[0]):
    for j in range(M.shape[1]):
        M[i,j] = (b-a)*random.random()+a       # random.random() genera un único valor aleatorio
print(M)
```

```python
M2 = (b-a)*np.random.rand(3,5)+a
print(M)
```

```python
arreglo = np.random.rand(3,5)
print(arreglo)

# Notar que los operadores suma y multiplicación se aplican sobre cada elemento del arreglo
M2 = (b-a)*arreglo+a
print('\n',M2)
```

Muchas de las funciones de la librería `math` son **funciones escalares**, es decir, están diseñadas para operar sobre valores numéricos individuales.

`Numpy` contiene las **versiones vectorizadas** de estas funciones, por ejemplo:
- math.sin --> numpy.sin
- math.exp --> numpy.exp

A diferencia de las funciones escalares, las versiones vectorizadas pueden operar elemento a elemento sobre arreglos $n$-dimensionales.

#### **Función escalar**

* La **función sigmoide** es una función matemática con forma de “S” que se usa mucho en áreas como el aprendizaje automático, la ciencia de datos, estadística y biología.
$$
\sigma(x) = \frac{1}{1 + e^{-x}}
$$
Esta función no está incluida en los módulos math o numpy. Por tanto, vamos a crear la función sigmoide:

```python
# Creemos la función sigmoide escalar (usando math)
def sigmoid(z):
    return 1. / (1. + math.exp(-z))
```

```python
plt.plot(x,R)
plt.grid()
plt.xlim([-5,5])
```

* Esta herramienta de conveniencia que transforma una función que acepta un solo escalar en una que puede aceptar un array de NumPy
* Sin embargo, es vital entender que `np.vectorize` no es verdadera vectorización a nivel de hardware.

### **9.3.5. Broadcasting**
El **broadcasting** es una de las características más potentes de NumPy. Es lo que permite realizar operaciones aritméticas entre arrays que tienen diferentes formas (shapes).

Normalmente, para operar con dos matrices, estas tendrían que ser del mismo tamaño. El **broadcasting** rompe esa regla siguiendo un conjunto de normas lógicas para "**estirar**" el array más pequeño de modo que coincida con el más grande.

```python
np.random.seed(2)

A0 = np.random.randint(0,4,(4,3))
print(A0)
print(f"Tamaño de A0 = {A0.shape}\n")

A1 = np.ones((4,1), dtype= int)

print(A1)
print(f"Tamaño de A1 = {A1.shape}")
```

```python
print(A0 + A1)
```

Para saber si dos arrays son compatibles, NumPy compara sus dimensiones de derecha a izquierda (empezando por la última). Dos dimensiones son compatibles si:

 - Son iguales.

- Una de ellas es 1.

```python
np.random.seed(2)

Ak = np.random.randint(0,4,(5,4))
print(Ak)
print(f"\nTamaño de Ak = {Ak.shape}\n")

media_filas = np.mean(Ak, axis=1)
print(media_filas)
print(f"\nTamaño de media_filas = {media_filas.shape}")
```

```python
# Los arreglos no son compatibles porque media_filas no tiene segunda dimensión
Akm = Ak - media_filas
print(media_filas)
print(Akm)
```

```python
# Convertimos el arreglo 'media_filas' de dimensiones (5,) a (5,1)
media_filas = np.resize(media_filas,(5,1))
print(media_filas)
print(f"\nTamaño de media_filas = {media_filas.shape}")
```

```python
Akm = Ak - media_filas
print(Akm)
```

## **9.4. Ejemplo: Distancia euclidiana**


Supongamos dos arreglos A y B de tamaño $m \times 2$, donde $m$ es un número de puntos y $2$ es la dimensión de los puntos, es decir, 2D para este ejemplo.

Vectorizar la distancia Euclídea entre cada punto 2D en A y su correspondiente punto 2D en B. El resultado debe ser un arreglo de dimensión $m$.

```python
np.random.seed(2)

m = 5000

A = np.random.randint(1000, size=(m,2))
B = np.random.randint(1000, size=(m,2))
#B = np.random.rand(m, 2)
```

```python
A
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

plt.figure(figsize=(3,3))
plt.plot(A[:,0], A[:,1], 'r*')
plt.plot(B[:,0], B[:,1], 'b.')
plt.show()
```

* Crear una función **escalar** para calcular la distancia entre dos puntos:

$distAB = \sqrt{ (x_2-x_1)^2 +(y_2-y_1)^2} $

```python
def distAB(ax,ay,bx,by):
  return math.sqrt((ax-bx)**2+(ay-by)**2)
```

```python
%%timeit
R = []
for k in range(A.shape[0]):
  R.append(distAB(A[k,0],A[k,1],B[k,0],B[k,1]))
```

* Vectorizar la función `distAB`:



```python
distAB_v = np.vectorize(distAB)
```

```python
%%timeit
distAB_v(A[:,0],A[:,1],B[:,0],B[:,1])
```

* Construir la función vectorizada `distAB_np`:

```python
def distAB_np(ax,ay,bx,by):
  return np.sqrt((ax-bx)**2+(ay-by)**2)
```

```python
%%timeit
distAB_np(A[:,0],A[:,1],B[:,0],B[:,1])
```
