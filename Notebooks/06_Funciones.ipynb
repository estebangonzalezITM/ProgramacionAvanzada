# **6. Funciones**

En Python, una **función** es un bloque de código reutilizable que realiza una tarea específica.

Las funciones permiten organizar mejor los programas, evitar repetir instrucciones y dividir problemas complejos en partes más pequeñas.

Pueden recibir datos de entrada llamados parámetros, procesarlos mediante distintas instrucciones y, si es necesario, devolver un resultado usando la palabra clave `return`.

## **6.4. Uso de**  `*args` **y** `**kwargs`

`*args` **y** `**kwargs` se usan para guardar datos cuando no se sabe cuantos datos (argumentos) va a recibir una función, o cuando hay datos opcionales.

* En realidad no tienes porque usar los nombres `args` o `kwargs`, ya que se trata de una mera convención entre programadores.

* Sin embargo lo que si que tienes que usar es el asterisco simple `*` o doble `**`. Es decir, podrías escribir `*datos` o `**datos` (o cualquier palabra) y el resultado será el mismo.

https://python-intermedio.readthedocs.io/es/latest/args_and_kwargs.html

```python
def test_var_args(*argv):
    n=0
    for i in argv:
        print("argumentos de *argv:", i)
        n += 1
    print(f'\nLa función recibió {n} argumentos')
    print(argv)
    print(type(argv))
```

```python
test_var_args('python', 'foo', 'bar')
```

```python
test_var_args(1 , 10 , 100 , 1 , 20 , 5 , 1 , 20 , 5)
```

```python
test_var_args()
```

* Como acabamos de ver, los datos guardados con `*args` se guardan como una tupla.
* Una misma función puede recibir datos "normales" y datos `*args` o `*kwargs`.

```python
def test_args(dato1=0, *argv):
    print("Primer argumento normal:", dato1)
    print(dato1)
    print(type(dato1))
    print(f'Los demás argumentos{argv}')
    print(type(argv))
```

```python
test_args('python', 'foo', 'bar')
```

```python
test_args(1,10,100,True)
```

```python
def test_args2(*datos):
    suma = sum(datos)
    print (f'Los datos ingresados son {datos}')
    return 'Los datos ingresados suman '+ str(suma)

test_args2(1,10,100)
```

La diferencia fundamental entre `*args` y `**kwargs` radica en cómo empaquetan los datos que sobran al llamar a una función.

* Si `*args` es una "bolsa" de valores, `**kwargs` es un "contenedor etiquetado", que permite pasar argumentos con un nombre asociado (llave=valor).

* `**args` guarda los datos en una tupla y `**kwargs` en un diccionario

```python
def test_kwargs(**elements):
    print(elements)
    print(type(elements))

test_kwargs(nombre="Covadonga", apellido = "Verdugo", edad1 = 37 )
# La estructura dque deben llevar los datos es:           llave1=valor1     ,    llave2=valor2
```

```python
def Hello(**world):
    print(world)
    print(type(world))

Hello(e1="Hola", e2 = "Mundo", e3 = 2026)
```

* **NOTA:** Al llamar una función que usa `**kwargs` <font color="red" size=3>__la llave no debe comenzar con un número__</font>, ya que python asume que se está creando una variable, y por tanto arrojará un error


```python
Hello(e1="Hola", e2 = "Mundo", e3 = 'tres')
```

### ¿Para qué se usan?

`*args` es útil para funciones matemáticas que reciben muchos números o para imprimir mensajes variables.

`**kwargs` es el estándar en librerías de visualización (como Matplotlib) o de Machine Learning (como Scikit-Learn). Se usa para pasar configuraciones y parámetros (como el color, el tamaño de la fuente, el número de neuronas, etc.) de forma opcional.

## **6.5. Valores de retorno**

Una función puede retornar un mensaje (similar que un print) o un valor (que puede usarse a posteriori). En cuyo caso utilizamos la palabra reservada __return__.

 - Sintaxis:
    
__return__ objeto_a_retornar


```python
def cuad(x):
    return x**2 #se retorna el valor x al cuadrado

a = 15

Respuesta = cuad(a)

print(Respuesta)
print(type(Respuesta))
print(cuad(a))
Respuesta/2
```

* Si solo se imprime (`print`), la función muestra el resultado, pero tú se puede usar ese número para otra operación matemática fuera de ella.

* Si se usa `return`, la función "entrega" resultado para que puedas guardarlo en una variable y seguir trabajando con él.

```python
def cuad2(x):
  print(x**2)

Rta = cuad2(a)
print(type(Rta))
print(Rta)
Rta/2
```

* Una función con valor de retorno, equivale en el momento del llamado, al tipo de dato que retorna. Es decir, una función que devuelve o retorna un valor, soporta todas las operaciones del tipo de dato que retorna.

```python
import math

x = 4
y = math.sqrt(cuad(x) + 9) #estamos utilizando cuad(x) como un flotante

print(y)
```

* Desempaquedatos extendido de iterables

```python
def ejemplo(*args):
    num1, num2, *_ = args
    # Esta línea asigna los dos primeros por posición a las variables num1 y num2.
    # Todo lo demás se "desecha" (se almacenan en una tupla)
    suma = num1 + num2
    print(f'Los números ingresados son {args}')
    return suma

ejemplo(1,2.0,3,4,5,6)
    #num1, num2, *_ = args es equivalente a hacer: num1 = args[1] y num2 = args[2]
```

* Si es necesario utilizar todos los elementos dentro de `args` o `kwargs`, se puede usar un ciclo

```python
def ejemplo(*args):
    sumaTotal = 0
    for num in args:
      sumaTotal += num
    print(f'Los números ingresados son {args}')
    return sumaTotal

ejemplo(1,2.0,3,4,5,6)
```

## **6.6. Múltiples valores de retorno**

En caso de que una funcion deba retornar más de un valor, podemos:

* Retornar los elementos separados por coma `( , , ... )`
* Agregar los elementos a retornar en una estructura de datos tipo lista `[ , , ... ]`
* Agregar los elementos a retornar en una estructura de datos tipo diccionario `{ k1:v1 , k2:v2 , ... }`





***
### **6.6.1. Retornar los elementos separados por coma, lo que crea una tupla `( , , ... )`**
En este caso, Python se encarga de agregar automáticamente los elementos en una tupla.

```python
## definir una función que reciba dos números como parámetros, debemos retornar
## la suma, la resta, el producto y la división de los primeros dos números

def ejemplo1(*args):
    num1, num2, *_ = args
    suma = num1+num2
    resta = num1-num2
    producto = num1*num2
    division = num1/num2
    print(f'Los números ingresados son {args}')

    return suma, resta, producto, division

ejm = ejemplo1(1,2,3,4,5,6)

print(f'ejm_tupla = {ejm}')
print(type(ejm))
```

***
### **6.6.2.  Agregar los elementos a retornar en una estructura de datos tipo lista `[ , , ... ]`**

```python
def ejemplo2(*args):
    num1, num2, *_ = args
    suma = num1+num2
    resta = num1-num2
    producto = num1*num2
    division = num1/num2
    print(f'Los números ingresados son {args}')

    resultados = [suma, resta, producto, division]

    return resultados

ejm = ejemplo2(1,2,3,4,5,6)
print(f'ejm_lista = {ejm}')
print(type(ejm))
```

```python
def ejemplo3(*args):
    num1, num2, *_ = args
    suma = num1+num2
    resta = num1-num2
    producto = num1*num2
    division = num1/num2
    print(f'Los números ingresados son {args}')

    return [suma, resta, producto, division]

ejm = ejemplo3(1,2,3,4,5,6)
print(f'ejm_lista = {ejm}')
print(type(ejm))
```

***
### **6.6.3.  Agregar los elementos a retornar en una estructura de datos tipo diccionario `{ k1:v1 , k2:v2 , ... }`**


```python
def ejemplo3(*args):
    num1, num2, *_ = args
    suma = num1+num2
    resta = num1-num2
    producto = num1*num2
    division = num1/num2
    print(f'Los números ingresados son {args}')

    return {"s":suma, "r":resta, "p":producto, "div":division}

ejm = ejemplo3(1,2,3,4,5,6)
print(f'ejm_lista = {ejm}')
print(type(ejm))
```

```python
a = 3

L = ejemplo3(1,a)
print(type(L))
print(L)
print(L["r"])
```

### **6.6.4.   También se puede usar con `**kwargs`**:

```python
## definir una función que reciba dos números como parámetros, debemos
## retornar la suma, la resta, el producto y la división de los
## dos números

def ejemplokwargs(**kwargs):
    num1 = kwargs['e1']      # Obliga a que el usuario entregue un argumento llamado exactamente e1
    num2 = kwargs['e2']      # Obliga a que el usuario entregue un argumento llamado exactamente e2
    suma = num1+num2
    resta = num1-num2
    producto = num1*num2
    division = num1/num2
    print(f'El diccionario ingresado es {kwargs}')
    lista = [suma, resta, producto, division]

    return lista

ejm = ejemplokwargs(e1=1,e2=2,e3=2,llave='52')

print(f'ejm_lista = {ejm}')
#print(type(ejm))
```

## **6.8. Alcance (Scope) de las variables**

Analizar el siguiente ejemplo:

```python
a=3
b=1

def suma(a):
  print(f'Dentro de la función a = {a}, b = {b}')
  return a+b

print(suma(1))

print(f'Fuera de la función a = {a}, b = {b}')
```

* Las variables `a` (a = 3) y `b` (b = 1) existen en el cuerpo principal del programa.

* Dentro de la función, la variable `a` es **Local** (parámetro). Al definir `def suma(a):`, creas una nueva variable `a` que solo existe dentro de la función.

* A diferencia de `a`, la variable `b` no está definida como parámetro ni se crea dentro de la función. Como la función no la encuentra dentro de la función, expande su búsqueda al ámbito **global**. Allí encuentra `b = 1` y utiliza ese valor.

***
### **6.8.1. Localidad y globalidad de una variable**

 * Una variable puede ser **local** o **global** respecto a una función, i.e., el valor de la variable puede existir de forma interna o externa a la función.
 * Por definición, todo parámetro de una función es **local** a la misma.

```python
a = 3
b = 0
c = 10 #c es una variable por fuera de la funcion suma, esto es, c es global

def suma(a, b):
    c = a + b #la variable c es local respecto a la funcion suma
    print("Valor de c dentro de la funcion: ", c)
    return c

suma(a, b)
print("Valor de c por fuera de la funcion: ", c)
```
