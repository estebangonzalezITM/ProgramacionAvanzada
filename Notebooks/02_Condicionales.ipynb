# **2.2. Operadores Relacionales en Python**


- Se usan para establecer relación entre valores (variables)
- Producen un resultado booleano (True o False):

Una expresión Booleana se puede definir como toda expresión que retorna un valor falso o verdadero:
     
$ expresion \rightarrow [True, False] $

- Todos tienen el mismo nivel de precedencia (prioridad), pero son de menor prioridad que los operadores aritméticos

| Operador | Operación |
| --- | --- |
| > | Mayor que |
| < | Menor que |
| >= | Mayor o igual que |
| <= | Menor o igual que |
| != | Diferente  |
| == | Igual |

```python
# Celda para ejemplos de operadores relacionales

a = 10
b = 3
```

```python
variable = a >= b

print(variable)
print(type(variable))
```

```python
# Los operadores se pueden encadenar (equivalente a usar el operador lógico AND)
print(0 < a <= 10)
```

```python
variable2 = 0 < a < 5
print(variable2)
```

# **2.4. Estructura condicionales**

En Python contamos con tres elementos condicionales de la siguiente forma:

**2.4.1. if**(_expresion booleana_)**:** --> Si se cumple la expresión booleana.

**2.4.2. else:** --> en caso contrario.

**2.4.3. elif**(_expresion booleana_)**:** --> Si no se cumple la expresión anterior y se cumple esta. Viene a reemplazar el `if` anidado de un nivel.

## **2.4.1. IF**


Si la condición que pones al lado es verdadera (True), el código dentro de él se ejecuta. Si no, se lo salta.

<br>

<font color="#130eb3" size=5>__if__</font> <font size=5>(condición)</font> <font color="#130eb3" size=5>__:__</font>

$\qquad$ instrucción 1
    
$\qquad$ instrucción 2
    
$\qquad$ ...

$\qquad$ instrucción n


```python
#Ejemplo if

edad = float(input("Ingrese su edad en años: "))
```

```python
if(edad >= 18):
    print("Eres mayor de edad")
    meses = edad*12
    print(f"Tienes aproximadamente {meses:.0f} meses")
    print("instruccion 4")

print("Esto está fuera del if")
```

## **2.4.2. IF - ELSE**

Si no se cumple la condición del IF (o cualquier otra condición anterior), se ejecuta automáticamente. Es opcional, pero útil para cerrar el ciclo.

<br>

<font color="#130eb3" size=5>__if__</font> <font size=5>(condición)</font> <font color="#130eb3" size=5>__:__</font>

$\qquad$ instrucción 1
    
$\qquad$ instrucción 2
    
$\qquad$ ...
    
$\qquad$ instrucción n
    
<font color="#130eb3" size=5>__else:__</font>

$\qquad$ instrucción sino 1
    
$\qquad$ instrucción sino 2
    
$\qquad$ ...
    
$\qquad$ instrucción sino n

No lleva ninguna condición porque se ejecuta automáticamente si ninguna de las condiciones anteriores fue verdadera. Es opcional, pero útil para cerrar el ciclo.

```python
#Ejemplo if - else

edad = float(input("Ingrese su edad en años: "))

if edad >= 18:
    print("Eres mayor de edad")
else:
    print("No eres mayor de edad")

print("Esto es fuera del if-else")
```

## **2.4.3. ELIF (ELSE - IF )**

Se usa cuando quieres revisar una segunda, tercera, cuarta... condición(es), pero solo si las anteriores fallaron. Puedes tener tantos elif como necesites.

<br>

<font color="#130eb3" size=5>__if__</font> <font size=5>(condición)</font> <font color="#130eb3" size=5>__:__</font>

$\qquad$ instrucción 1
    
$\qquad$ instrucción 2
    
$\qquad$ ...
    
$\qquad$ instrucción n
    
<font color="#130eb3" size=5>__elif__</font> <font size=5>(condición2)</font> <font color="#130eb3" size=5>__:__</font>

$\qquad$ instrucción condición2 1
    
$\qquad$ instrucción condición2 2
    
$\qquad$ ...
    
$\qquad$ instrucción condición2 n
    
<font color="#130eb3" size=5>__else:__</font>

$\qquad$ instrucción sino 1
    
$\qquad$ instrucción sino 2
    
$\qquad$ ...
    
$\qquad$ instrucción sino n

```python
#Ejemplo elif - else

edad = float(input("Ingrese su edad en años: "))

if (edad < 18 and edad >= 0):
  print("No eres mayor de edad")
elif (edad < 0):
  print("Error")
elif (edad >= 60):
  print("Eres un adulto mayor")
else:
  print("Eres mayor de edad")
```

```python
# Ejemplo if-elif-else
# Solicitar coordenadas de un punto y decir en qué cuadrante del plano cartesiano se encuentra
# (x, y)

x = float(input("ingrese el valor de x: "))
y = float(input("ingrese el valor de y: "))

if (x>0):
    if (y<0):
        print("El punto está en el cuadrante 4")
    else:
        print("El punto está en el cuadrante 1")
else:
    if (y<0):
        print("El punto está en el cuadrante 3")
    else:
        print("El punto está en el cuadrante 2")
```

## **2.4.4. Inline IF**



Se suele utilizar para asignar el resultado de una selección condicional a una variable. La estructura de un **inline if** es la siguiente:


* $(\text{Valor si es True}) \ \color{blue}{\textbf{if}} \ (\text{Condición}) \ \color{blue}{\textbf{else}} \ (\text{Valor si es False})$

* $\text{Variable} = (\text{Valor si es True}) \ \color{blue}{\textbf{if}} \ (\text{Condición}) \ \color{blue}{\textbf{else}} \ (\text{Valor si es False})$




```python
num = int(input("Ingrese un número entero: "))

# Supongamos la estructura IF - ELSE:
if num%2 == 0:
  print('Numero par')
else:
  print('Numero impar')
```

```python
#Inline if
print('Numero par') if num%2 == 0 else print('Numero impar')
```

```python
'''
Si x es menor que 10, asigne x al cuadrado a la variable msg,
en caso contrario, asigne el valor 1 sobre x a la variable msg
'''
x = 20

if (x < 10):
    msg = x**2
else:
    msg = 1/x

print(msg)
```

```python
x = 20

msg = x**2 if x < 10 else 1/x

print(msg)
```

# **2.7. Estructura match**

Es una herramienta de Coincidencia de Patrones Estructurales (Structural Pattern Matching).

El símbolo _ (guion bajo) actúa como un comodín que atrapa cualquier valor que no haya coincidido con los casos anteriores (similar al else).

**Ejemplo:** Saludar en diferentes idiomas

```python
idioma = input("Ingrese el idioma (es, en, fr): ")

match idioma:
    case "es":
        print("Hola!")
    case "en":
        print("Hello!")
    case "fr":
        print("Salut!")
    case _:
        print("Idioma no soportado")
```

También puede resolverse usando IF

```python
idioma = input("Ingrese el idioma (es, en, fr): ")
if idioma == 'es':
    print("Hola!")
elif idioma == 'en':
    print("Hello!")
elif idioma =='fr':
    print("Salut!")
else:
    print("Idioma no soportado")
```

**Ejemplo:** Clasificar si un punto esta en el origen o en algun eje.

```python
coordenada = (1, 0) #Cambiar valores
match coordenada:
    case (0, 0):
        print("Coordenada en origen")
    case (x, 0):
        print(f"Coordenada en eje x: {x}")
    case (0, y):
        print(f"Coordenada en eje y: {y}")
    case (x, y):
        print(f"Coordenada en: {x}, {y}")
    case _:
        print("Error")
```

**¿En qué contexto es mejor que un IF-ELIF?**

A diferencia de los condicionales tradicionales, match es superior mejor cuando trabaja con datos complejos como listas, diccionarios o clases.
