## **1.4. Objetos en Python**

**Todo en Python es un objeto**, esto significa que todo dato tiene:

* Atributos.
* Funciones (métodos).

Podemos consultar los métodos de un tipo de datos, accediendo a la documentación de Python. Por ejemplo, para los Strings: https://docs.python.org/3.7/library/string.html

````python
numComplejo = 4 + 3j
print(numComplejo)
````

Los **atributos** son las características o el estado del objeto, por ejemplo un número complejo tiene parte real e imaginaria

````python
print(numComplejo.real)          # .real es un atributo del tipo de datos Complex. Atributo: 4
print(numComplejo.imag)          # .imag es un atributo del tipo de datos Complex. Atributo: 3
````

Los **métodos** son los comportamientos o las funciones internas asociadas a un objeto.

````python
print(numComplejo.conjugate())   # .conjugate es un método del tipo de datos Complex.
````

- Los **atributos** son lo que el objeto es / tiene (Estado)
- Los **métodos** son lo que el objeto hace (Comportamiento)

### **1.6.1. Palabras integradas (Built-in) y palabras reservadas (Keyword)**

Todo lenguaje de programación tiene palabras que son de uso privativo del mismo, es decir, que **no es conveniente** (integradas) y en algunos casos es **imposible** (reservadas)utilizarlas, por ejemplo:

- sqrt, abs, etc.
- sin, cos, tan, etc.
- for, while, if, else, etc.

Algunas palabras no pueden/deben ser utilizadas en la definición de elementos dentro del programa, pues Python las conserva reservadas para ciertas funciones. Algunos casos son: *if, for, while, import, from, True, False, etc*: https://www.programiz.com/python-programming/keywords-identifier

````python
r = float(input("Ingrese el radio "))

print(r)
print(type(r))

b = pi*r**2

print("El área del circulo con radio:",r," es:",b)

print("El área del circulo con radio: %.2f es: %.5f"%(r,b)) #caracteres especiales printf

print("El área del circulo con radio: {:.1f} es: {:.3f}".format(r,b))

print(f"El área del circulo con radio: {r} es: {b:.5g}")

print(f"El área del circulo con radio: {r} es: {b:.5f}")
````

**NOTA:**

La diferencia fundamental entre usar `f` y `g` radica en qué se está contando para formatear o redondear el número:

- `f` (Fixed-point / Punto Fijo):

Cuenta el **número de decimales** (dígitos a la derecha del punto).

- `g` (General / Cifras Significativas):

Cuenta el total de **cifras significativas** (dígitos importantes en total, antes y después del punto).

````python
a = 1.232432556234234324
b = 0.0012344352565352434434

print("%.2f"%(a))
print("%.2g"%(a))

print("%.2f"%(b))
print("%.2g"%(b))
````
